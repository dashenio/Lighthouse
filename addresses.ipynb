```python
#import things
import pandas as pd
# pd.set_option('max_columns', 200) mostra todas as colunas da tabela
```

```python
#dataset
df = pd.read_csv('raw_data/addresses.csv')
df.info()
# df.shape ---> mostra quantas linhas e colunas tem a tabela 
```

```python
# Começar a limpar
df = df.set_index('id') # coloca a coluna id como index da tabela
# df.dtypes mostra o tipo de dados de cada coluna
df
```

```python
# encontrando missing values
df.isna().sum() # encontra todos os campos onde falta valor (no caso, encontra os valores NaN)
```

```python
df.duplicated() # avalia se o registro é duplicado
# df.loc[df.duplicated()] mostra os registros duplicados se houver
# df.loc[df.duplicated(subset=['postal_code'])]
```

```python
df.loc[df.duplicated(subset=['street'])] # identifica se valores de colunas específicas estão duplicados
```

```python
# tirar o hífen do CEP
cep = df['postal_code']
df['postal_code'] = cep.str.replace('-', '', regex=False) 
df['postal_code'].tail()
```

```python
# ruas = df[df['street'] == 'Ladeira de da Luz']
cidades = df.loc[df['state'] == 'SP', 'city'] # cidades por estado
cidades
```

```python
# Tirar o 'de da' dos nomes das cidades
df.query("city.str.contains('de da')")
df = df.copy()
df['city'] = df['city'].str.replace('de da', 'da')
#df.query("city.str[0].str.islower() == True and state == 'SP'")
df
```
